```python
import cmdstanpy as csp
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from cmdstanpy import install_cmdstan

install_cmdstan(overwrite=True, compiler=True, cores=16)
```

```python
df = pd.read_csv("../no2.csv")
df["t_raw"] = df.index + 1
df["t"] = (df["t_raw"] - df["t_raw"].mean()) / df["t_raw"].std()
```

```python
data_dict = {
    "n": len(df),
    "t": df["t"],
    "y": df["no2"],
    "J_year": 2,
    "J_week": 1
}
```

```python
model = csp.CmdStanModel(stan_file="../models/piecewise_linear.stan", force_compile=True)
```

```python
fit = model.sample(
    data=data_dict,
    chains=4,
    parallel_chains=4,
    seed=42,
    iter_warmup=1000
)
```

```python
fit.summary()
```
