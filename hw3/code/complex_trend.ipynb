```python
import cmdstanpy as csp
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from cmdstanpy import install_cmdstan

install_cmdstan(overwrite=True, compiler=True, cores=16)
```

```python
df = pd.read_csv("../no2.csv")
df["t_raw"] = df.index + 1
df["t"] = (df["t_raw"] - df["t_raw"].mean()) / df["t_raw"].std()
```

```python
data_dict = {
    "n": len(df),
    "t": df["t"],
    "y": df["no2"],
    "S": 3,
    "period": [365.25, 1826, 3652.5]
    }
```

```python
model = csp.CmdStanModel(stan_file="../models/complex_trend.stan", force_compile=True)
```

```python
fit = model.sample(
    data=data_dict,
    chains=4,
    parallel_chains=4,
    seed=42,
    iter_warmup=500,
    show_console=True
)
```

```python
fit.summary()
```

```python
import arviz as az
# Convert fit to InferenceData
idata = az.from_cmdstanpy(fit)

# 1. Check Traceplots
# Look for "fuzzy caterpillars" - chains should overlap and not drift
az.plot_trace(idata)

# 2. Check Stats (R-hat and ESS)
summary = az.summary(idata)
print(summary)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Extract the means of the parameters
params = fit.stan_variables()
a_hat = params['alpha'].mean()
b_hat = params['beta'].mean()
b_cos_hat = params['beta_cos'].mean(axis=0) # Vector of size S
b_sin_hat = params['beta_sin'].mean(axis=0) # Vector of size S

# 2. Reconstruct the components
t_vals = df['t'].values
y_vals = df['no2'].values
periods = data_dict['period']
omega = 2 * np.pi / np.array(periods)

# Start with trend: alpha + beta * t
mu_pred = a_hat + b_hat * t_vals

# Add the S long-term seasonalities
for s in range(len(periods)):
    mu_pred += (b_cos_hat[s] * np.cos(omega[s] * t_vals) + 
                b_sin_hat[s] * np.sin(omega[s] * t_vals))

# 3. Visualization
plt.figure(figsize=(14, 7))

# Plot raw standardized data
plt.scatter(t_vals, y_vals, color='gray', s=5, alpha=0.2, label='Standardized NO2 Data')

# Plot the combined Prediction (Trend + Long Seasons)
plt.plot(t_vals, mu_pred, color='crimson', label='Decadal Model (Trend + Fourier)', linewidth=2)

# Plot just the Linear Trend to see the difference
plt.plot(t_vals, a_hat + b_hat * t_vals, color='blue', linestyle='--', alpha=0.6, label='Underlying Linear Trend')

plt.title("Long-term NO2 Modeling: Linear Trend + Decadal Fourier Cycles")
plt.xlabel("Standardized Time")
plt.ylabel("Standardized Concentration")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
