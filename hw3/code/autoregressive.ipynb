```python
import cmdstanpy as csp
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from cmdstanpy import install_cmdstan

install_cmdstan(dir="C:/cmdstan", overwrite=True, compiler=True)
```

```python
model = csp.CmdStanModel(stan_file="..\models\\autoregressive.stan", force_compile=True)
```

```python
df = pd.read_csv("../no2.csv")
df["day"] = df.index + 1

df["day"] = df["day"] - df["day"].mean()
max_day = df["day"].max()
df["day"] /= max(abs(df["day"]))
```

```python
seasonality = 7 / max_day
```

```python
stan_data = {
    "n": len(df),
    "t": df["day"],
    "y": df["no2"],
    "k": 2,
    "omega": np.array([2*np.pi/seasonality, 2*np.pi/(365.25 / max_day)])
}
```

```python
fit = model.sample(
    data=stan_data,
    chains=4,
    parallel_chains=4,
    iter_warmup=1000,
    iter_sampling=1000,
    seed=42,
)
```
